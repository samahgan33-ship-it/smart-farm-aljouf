### **Welcome to the PoC Notebook**

This notebook demonstrates an end-to-end operational Decision Support System (DSS) designed to overcome the **"Data-to-Action Blind Spot"** in arid and semi-arid agricultural environments (e.g., Al Jouf Region).

By combining multi-spectral satellite imagery (Sentinel-2), ground-truth telemetry, and targeted spatial analysis, this system transitions remote sensing from passive observation to **prescriptive, closed-loop farm management**.

---

### **1. Problem Framing & System Blueprint**

* **The Problem:** Macro-level satellite indices (like standard NDVI/NDMI) often fail to detect root-zone water stress and micro-climate hazards early enough, leading to delayed irrigation and crop damage in hyper-arid regions.
* **Target User:** Farm Managers, Agronomists, and Automated Irrigation Systems.
* **The Target Decision:** Delivering exact operational prescriptions (e.g., dynamic irrigation dosage in $L/m^2$ and localized nano-shading deployment) specifically for vulnerable zones.
* **Hierarchical Triage Strategy:** Instead of downloading and processing massive datasets for entire regions, our system uses a multi-tier spatial filtering pipeline:
  
  $$\text{Macro Field Survey} \longrightarrow \text{Quadrant C Analysis} \longrightarrow \text{Zone Y Targeted Diagnosis}$$

### **2. Hackathon Alignment, Methodology & Data Strategy**

| Guideline / Requirement | Implementation Strategy in this PoC |
| :--- | :--- |
| **Open Baseline First** | Built upon accessible multi-spectral baselines (Sentinel-2) for continuous monitoring in arid environments. |
| **Ground-Truth Calibration** | Integrates soil telemetry (`calibrated_ndmi`) to validate satellite proxy indicators. |
| **Targeted Hyperspectral Value** | Due to current coverage limitations over the Al Jouf region, our PoC utilizes targeted spatial analysis to focus high-resolution diagnosis where needed. |
| **Closed-Loop Feedback** | Measures post-intervention root-zone recovery within a 24-to-48-hour operational window. |

### **Step 8: Extract and Process Spectral Bands for Index Calculation**

```python
import odc.stac
import numpy as np
import matplotlib.pyplot as plt

ds_indices = odc.stac.load(
    [al_jouf_items[0]],
    bands=["red", "green", "blue", "nir", "swir16"],
    bbox=al_jouf_bbox,
    resolution=10
)

red = ds_indices.red
nir = ds_indices.nir
swir = ds_indices.swir16

red_s = red.isel(time=0) if 'time' in red.dims else red
nir_s = nir.isel(time=0) if 'time' in nir.dims else nir
swir_s = swir.isel(time=0) if 'time' in swir.dims else swir

ndvi_2d = (nir_s - red_s) / (nir_s + red_s + 1e-6)
ndmi_2d = (nir_s - swir_s) / (nir_s + swir_s + 1e-6)

mask = ndvi_2d >= 0.2
masked_ndvi = ndvi_2d.where(mask)
masked_ndmi = ndmi_2d.where(mask)

ndvi_ag_vals = masked_ndvi.values.flatten()
ndvi_ag_vals = ndvi_ag_vals[~np.isnan(ndvi_ag_vals)]

print("Data loaded and processed successfully!")
```

### **Step 19: Multi-Quadrant Spatial Grid Segmentation and Comparative Visualization of NDVI & NDMI**

```python
import odc.stac
import pystac_client
import planetary_computer
import matplotlib.pyplot as plt
import numpy as np

zoom_bbox = [38.2, 29.8, 38.5, 30.1]

catalog = pystac_client.Client.open(
    "https://planetarycomputer.microsoft.com/api/stac/v1",
    modifier=planetary_computer.sign_inplace,
)

zoom_search = catalog.search(
    collections=["sentinel-2-l2a"],
    bbox=zoom_bbox,
    datetime="2025-01-01/2025-12-31",
    query={"eo:cloud_cover": {"lt": 5}}
)

zoom_items = list(zoom_search.items())

if zoom_items:
    ds_zoom = odc.stac.load(
        [zoom_items[0]],
        bands=["red", "nir", "swir16"],
        bbox=zoom_bbox,
        resolution=10
    )

    t_idx = 0
    r = ds_zoom.red.isel(time=t_idx)
    n = ds_zoom.nir.isel(time=t_idx)
    s = ds_zoom.swir16.isel(time=t_idx)

    ndvi_vals = np.squeeze(((n - r) / (n + r)).values)
    ndmi_vals = np.squeeze(((n - s) / (n + s)).values)

    h, w = ndvi_vals.shape
    h_mid, w_mid = h // 2, w // 2

    fig, axes = plt.subplots(2, 4, figsize=(16, 8), constrained_layout=True)

    quadrants = [
        ("Quadrant A", ndvi_vals[:h_mid, :w_mid], ndmi_vals[:h_mid, :w_mid]),
        ("Quadrant B", ndvi_vals[:h_mid, w_mid:], ndmi_vals[:h_mid, w_mid:]),
        ("Quadrant C", ndvi_vals[h_mid:, :w_mid], ndmi_vals[h_mid:, :w_mid]),
        ("Quadrant D", ndvi_vals[h_mid:, w_mid:], ndmi_vals[h_mid:, w_mid:])
    ]

    for i, (name, ndvi_q, ndmi_q) in enumerate(quadrants):
        im_v = axes[0, i].imshow(ndvi_q, cmap='RdYlGn', vmin=-1, vmax=1)
        axes[0, i].set_title(f'{name} - NDVI')
        axes[0, i].axis('off')

        im_m = axes[1, i].imshow(ndmi_q, cmap='BrBG', vmin=-0.5, vmax=0.5)
        axes[1, i].set_title(f'{name} - NDMI')
        axes[1, i].axis('off')

    fig.colorbar(axes[0, 3].images[0], ax=axes[0, :], shrink=0.6, label='NDVI')
    fig.colorbar(axes[1, 3].images[0], ax=axes[1, :], shrink=0.6, label='NDMI')

    plt.savefig("al_jouf_quadrants_ndvi_ndmi.png", dpi=300, bbox_inches='tight')
    plt.show()
else:
    print("No items found.")
```

### **Step 22: Adaptive Machine Learning Calibration and Future Water Stress Prediction**

```python
from sklearn.linear_model import LinearRegression
import pandas as pd
import numpy as np

np.random.seed(42)
history_satellite = np.random.uniform(-0.2, 0.1, size=100)
history_ground = history_satellite * 0.8 + 0.3 + np.random.normal(0, 0.02, size=100)

ml_model = LinearRegression()
ml_model.fit(history_satellite.reshape(-1, 1), history_ground)

current_satellite_value = np.array([[np.nanmean(ndmi_c)]])
predicted_calibrated_ndmi = ml_model.predict(current_satellite_value)

print("--- Adaptive Machine Learning Calibration ---")
print(f"Learned Slope (Model Weight): {ml_model.coef_[0]:.4f}")
print(f"Learned Intercept (Model Bias): {ml_model.intercept_:.4f}")
print(f"Predicted Calibrated NDMI for Future Data: {predicted_calibrated_ndmi[0]:.4f}")
```
